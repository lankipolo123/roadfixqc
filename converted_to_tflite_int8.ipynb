```python
# ============================================================================
# YOLOv11 to TFLite INT8 Converter (3x Compression)
# ============================================================================

print("="*60)
print("🔧 YOLOv11 → TFLite INT8 Converter")
print("="*60)

# INSTALL
!pip install ultralytics tensorflow

import os
from google.colab import files
from ultralytics import YOLO
import shutil

# ============================================================================
# UPLOAD MODEL
# ============================================================================

print("\n📤 Upload your YOLOv11 model (.pt file)...")
uploaded = files.upload()

# Get uploaded filename
model_filename = list(uploaded.keys())[0]
print(f"✅ Uploaded: {model_filename}")

# ============================================================================
# LOAD MODEL
# ============================================================================

print(f"\n🔄 Loading model: {model_filename}")
model = YOLO(model_filename)
print("✅ Model loaded successfully!")

# ============================================================================
# CONVERSION 1: TFLite (FP32 - Full Precision)
# ============================================================================

print("\n🔧 Converting to TFLite FP32 (baseline)...")
model.export(format='tflite')
print("✅ TFLite FP32 created!")

# ============================================================================
# CONVERSION 2: TFLite (FP16 - Half Precision)
# ============================================================================

print("\n🔧 Converting to TFLite FP16 (2x smaller)...")
model.export(format='tflite', half=True)
print("✅ TFLite FP16 created!")

# ============================================================================
# CONVERSION 3: TFLite (INT8 - Quantized, 4x smaller)
# ============================================================================

print("\n🔧 Converting to TFLite INT8 (3-4x smaller, fastest)...")
model.export(format='tflite', int8=True)
print("✅ TFLite INT8 created!")

# ============================================================================
# ORGANIZE & DOWNLOAD
# ============================================================================

print("\n📦 Organizing files for download...")

# Create download folder
download_folder = '/content/TFLITE_MODELS'
os.makedirs(download_folder, exist_ok=True)

# Find all generated files
model_name = model_filename.replace('.pt', '')

tflite_files = {
    f'{model_name}_saved_model/{model_name}_float32.tflite': 'FP32_full_precision.tflite',
    f'{model_name}_saved_model/{model_name}_float16.tflite': 'FP16_half_precision.tflite',
    f'{model_name}_saved_model/{model_name}_int8.tflite': 'INT8_quantized.tflite',
}

# Copy and rename
for src_pattern, dst_name in tflite_files.items():
    # Try to find the file
    for root, dirs, files in os.walk('.'):
        for file in files:
            if file.endswith('.tflite'):
                src = os.path.join(root, file)

                # Determine type based on filename
                if 'int8' in file.lower() or 'int8' in src.lower():
                    final_name = f"{model_name}_INT8.tflite"
                elif 'float16' in file.lower() or 'fp16' in file.lower():
                    final_name = f"{model_name}_FP16.tflite"
                elif 'float32' in file.lower() or 'fp32' in file.lower():
                    final_name = f"{model_name}_FP32.tflite"
                else:
                    final_name = file

                dst = os.path.join(download_folder, final_name)

                # Copy if not already there
                if not os.path.exists(dst):
                    shutil.copy(src, dst)
                    print(f"✅ Copied: {final_name}")

# Get file sizes
print("\n📊 File Sizes:")
for file in os.listdir(download_folder):
    filepath = os.path.join(download_folder, file)
    size_mb = os.path.getsize(filepath) / (1024 * 1024)
    print(f"  {file}: {size_mb:.2f} MB")

# Create info file
with open(f'{download_folder}/CONVERSION_INFO.txt', 'w') as f:
    f.write("="*60 + "\n")
    f.write("TFLite Conversion Summary\n")
    f.write("="*60 + "\n\n")
    f.write(f"Original Model: {model_filename}\n\n")
    f.write("Generated Models:\n\n")

    for file in os.listdir(download_folder):
        if file.endswith('.tflite'):
            filepath = os.path.join(download_folder, file)
            size_mb = os.path.getsize(filepath) / (1024 * 1024)

            if 'INT8' in file:
                f.write(f"📱 {file} ({size_mb:.2f} MB)\n")
                f.write("   - BEST FOR MOBILE/FLUTTER\n")
                f.write("   - 3-4x smaller than original\n")
                f.write("   - Fastest inference\n")
                f.write("   - Minimal accuracy loss (~1-2%)\n\n")
            elif 'FP16' in file:
                f.write(f"📱 {file} ({size_mb:.2f} MB)\n")
                f.write("   - Good balance\n")
                f.write("   - 2x smaller than FP32\n\n")
            elif 'FP32' in file:
                f.write(f"💻 {file} ({size_mb:.2f} MB)\n")
                f.write("   - Full precision\n")
                f.write("   - Largest file\n\n")

    f.write("\n" + "="*60 + "\n")
    f.write("RECOMMENDED FOR FLUTTER: INT8 version\n")
    f.write("="*60 + "\n")

# Create ZIP
shutil.make_archive('/content/TFLite_Models_Complete', 'zip', download_folder)

print("\n" + "="*60)
print("📥 DOWNLOADING ALL TFLITE MODELS...")
print("="*60)
files.download('/content/TFLite_Models_Complete.zip')

print("\n✅ DONE! Use the INT8 version for Flutter!")
```
