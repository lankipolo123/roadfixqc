```python
# ================================
# 1. INSTALL
# ================================
!pip install ultralytics roboflow -q

# ================================
# 2. IMPORTS
# ================================
import os
import shutil
from ultralytics import YOLO
from roboflow import Roboflow
import torch
from google.colab import drive
from google.colab import files

# ================================
# 3. MOUNT DRIVE
# ================================
drive.mount('/content/drive')

# ================================
# 4. GPU CHECK
# ================================
if not torch.cuda.is_available():
    raise SystemExit("❌ GPU not enabled")

print("✅ GPU:", torch.cuda.get_device_name(0))

# ================================
# 5. DOWNLOAD DATASET
# ================================
rf = Roboflow(api_key="SzttdelfmuWaCwAz2N5u")
project = rf.workspace("dequillaprojects").project("roadfix")
version = project.version(3)
dataset = version.download("yolov11", location="/content/RoadFixDataset")

# ================================
# 6. LOAD MODEL
# ================================
model = YOLO("yolo11s.pt")

# ================================
# 7. TRAIN
# ================================
model.train(
    data=dataset.location + "/data.yaml",
    epochs=150,
    imgsz=640,
    batch=-1,
    optimizer="AdamW",
    lr0=0.001,
    lrf=0.01,
    weight_decay=0.0005,
    device=0,
    workers=8,
    amp=True,
    cos_lr=True,
    patience=30,
    save_period=25,
    project="/content/drive/MyDrive/roadfix_training",
    name="roadfix_v4",
    exist_ok=True
)

print("✅ TRAINING DONE")

# ================================
# 8. ZIP + DOWNLOAD RESULTS
# ================================
training_path = "/content/drive/MyDrive/roadfix_training/roadfix_v4"
zip_path = "/content/roadfix_results"

shutil.make_archive(zip_path, 'zip', training_path)
files.download(zip_path + ".zip")

print("⬇️ Download started")
```
